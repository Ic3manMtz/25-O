# Bibliotecas

```python
import numpy as np
import pandas as pd 

from sklearn.model_selection import KFold # Pra validación cruzada
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC

from sklearn.metrics import precision_score, confusion_matrix

# Para usar datos de CSVs, Ordinal es para datos y Label para clases
from sklearn.preprocessing import OrdinalEncoder, LabelEncoder

from sklearn.model_selection import train_test_split

import tensorflow as tf
```

# Lee datos

```python
df = pd.read_csv('agaricus-lepiota.csv')
df.head(7)
```

# Haz X(datos) y y(Clases)

```python
Xn = df.drop(columns=['class'], axis=1)
yn =df['class']

print(Xn.head(2), yn.head(2))
```

# Codificar como números

```python
encX = OrdinalEncoder() # Para datos
encX.fit(Xn)
X = encX.transform(Xn)

ency = LabelEncoder() # Para clases
ency.fit(yn)
y = ency.transform(yn)

print(X[0:2], y[0:2])
```

# Divide en datos de entrenamiento

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.7, random_state=7)
```

# Crea y entrena modelo

```python
X.shape
```

```python
modelo1 = tf.keras.models.Sequential([
    tf.keras.layers.Flatten(input_shape=(22,1)),
    tf.keras.layers.Dense(128, activation='relu'),
    tf.keras.layers.Dropout(0.2),
    tf.keras.layers.Dense(2,activation='softmax')
])

# Paso adicional: Compilación del modelo
modelo1.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)


modelo1.fit(X_train,y_train,epochs=30)
```

# Evalua modelo

```python
modelo1.evaluate(X_test,y_test)
```
